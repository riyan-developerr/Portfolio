```python
from openai import OpenAI
from PIL import Image
import base64
import io

client = OpenAI()

prompt = input("Enter your prompt: ")

result = client.images.generate(
    model="gpt-image-2",
    prompt=prompt,
    size="1024x1024",
    quality="medium",
    n=1,
    output_format="png"
)

image_base64 = result.data[0].b64_json
image_bytes = base64.b64decode(image_base64)

with open("generated_image.png", "wb") as f:
    f.write(image_bytes)

image = Image.open(io.BytesIO(image_bytes))
image
```

```python
#modules need for image generation and saving
from openai import OpenAI
from PIL import Image
import io
import base64
# import requests
#creating client
client=OpenAI()

#asking for prompt
prompt=input("Enter your prompt:")

request=client.images.generate(
    model="gpt-image-2",
    size="1024x1024",
    quality="low",
    n=1,
    prompt=prompt,
    output_format="png"
)
#output is in base64 text format
b64=request.data[0].b64_json
#converting from basee64 format into image(binary) format
image_bytes=base64.b64decode(b64)

with open("2_gen.png","wb") as f:
    f.write(image_bytes)
    
image=Image.open(io.BytesIO(image_bytes))
image.show()
```
