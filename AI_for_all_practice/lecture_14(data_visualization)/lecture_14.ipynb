```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("/content/seaborn_practice_ecommerce_sales.csv")
df.head()
```

```python
sns.histplot(df["revenue_usd"], kde=False)
```

```python
df.shape
```

```python
df.info()
```

```python
sns.displot(data=df,x="gender",height=6,aspect=2)
plt.show()
```

```python
sns.displot(data=df,x="month")
plt.show()
```

```python
sns.displot(data=df,x="profit_usd",bins=50,height=6,aspect=2,kde=True)
plt.show()
```

```python
sns.displot(data=df,x="revenue_usd",kind="hist",hue="gender",multiple="layer",bins=30)
plt.show()
```

```python
sns.displot(
    data=df,
    x="revenue_usd",
    hue="gender",
    kind="hist",
    multiple="stack",
    bins=30
)
plt.show()
```

```python
sns.displot(
    data=df,
    x="revenue_usd",
    hue="gender",
    kind="hist",
    multiple="dodge",
    bins=20
)
plt.show()
```

```python
sns.displot(
    data=df,
    x="revenue_usd",
    col="region",
    kind="hist",
    kde=True
)
plt.show()
```

```python
sns.relplot(data=df, x="ad_spend_usd", y="revenue_usd",kind="scatter",hue="gender")
plt.show()
```

```python
monthly_revenue = (
    df.groupby("order_date", as_index=False)["revenue_usd"]
    .sum()
)

monthly_revenue.head()
```

```python
sns.regplot(
    data=df,
    x="profit_usd",
    y="revenue_usd"
)

plt.show()
```
