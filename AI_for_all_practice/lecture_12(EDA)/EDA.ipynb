Question: which disease is most common?

```python
df["Disease"].value_counts()
```

```python
df.groupby("Disease")["Severity_Score"].mean().sort_values(ascending=False)
```

```python
df.groupby("Disease").agg({
    "Age": "mean",
    "Severity_Score": "mean",
    "Hospital_Days": "mean",
    "Treatment_Cost_USD": "mean"
}).sort_values("Severity_Score", ascending=False)
```
